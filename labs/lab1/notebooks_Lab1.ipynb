```python
# python code for solving a quardratic equation
import math
#solve ax^2+bx+c=0
def q_solve(a,b,c):
# compute the dicriminate D=b^2-4*a*c
  D=b**2 - 4*a*c 
  if D>=0:
    x1=(-b+math.sqrt(D))/(2*a)
    x2=(-b+math.sqrt(D))/(2*a)
    return x1,x2
  else:
    print("no real solution")
    return None, None
x1,x2=q_solve(1,4,1)
print(x1,x2)
x1,x2=q_solve(1,1,1)
```

```python
import math

def q_solve(a,b,c):
    D=b**2 - 4*a*c
    if D>=0:
        sq_D=math.sqrt(D)
    else:
        sq_D=math.sqrt(-D)*1j
    x1=(-b+sq_D)/(2*a)
    x2=(-b-sq_D)/(2*a)
    return x1,x2
x1,x2=q_solve(1,4,1)
print(x1,x2)
x1,x2=q_solve(1,1,1)
print(x1,x2)
```

```python
#solution where a,b,c are complex numbers 
import cmath

def q_solve(a,b,c):
    D=b**2 - 4*a*c
    sq_D=cmath.sqrt(D)
    x1=(-b+sq_D)/(2*a)
    x2=(-b-sq_D)/(2*a)
    return x1,x2
x1,x2=q_solve(1,4,1)
print(x1,x2)
x1,x2=q_solve(1,1,1)
print(x1,x2)
x1,x2=q_solve(1+2j,3-1j,2+1j)
print(x1,x2)
```
